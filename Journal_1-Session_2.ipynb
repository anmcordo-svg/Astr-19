```python
float_1 = 1.1
float_2 = 2.2

sum = float_1 + float_2

print(sum)
print(type(sum))
```

```python
integer_1 = 2
integer_2 = 1

dif = integer_1 - integer_2

print(dif)
print(type(dif))
```

```python
product = integer_1 * float_2

print(product)
print(type(product))
```
