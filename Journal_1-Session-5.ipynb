# Session 5 Prompt:
Write a Python program that writes out a table of the function sin(x) vs. x, 
where x is tabulated between 0 and 2pi with a thousand entries. Follow the basic Python program structure,
including a main program function and the if __name__ == “__main__” setup.


```python
import numpy as np

def main():
    x = np.linspace(0, 2 * np.pi, 1000)
    y = np.sin(x)
    grid = np.column_stack((x, y))

    print(f"    x      |     sin(x)")
    print("-" * 30)
    
    for i in range(len(x)):
        print(f"{x[i]:<10.5f} | {y[i]:>10.5f}")
    
if __name__ == "__main__":
    main()
```
