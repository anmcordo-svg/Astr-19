# NumPy
NumPy is a package for scientific computing in Python's coding library

- We are imported it because we need to use its features for sin and cos

```python
import numpy as np
```

# Sin
- The function that helps users calculate trigonometric sine for array elememts

```python
def calc_sin(x):
    return np.sin(x)
```

# Cos
- The function that helps users calculate trigonometric cosine for array elememts

```python
def calc_cos(x):
    return np.cos(x)
```

# Table
- This table contains the first 10 values for x, sin(x), and cos(x)

```python
x = np.linspace(0, 2 * np.pi, 1000)
sin_x = calc_sin(x)
cos_x = calc_cos(x)
```

```python
print(f"     x     |   sin(x)   |  cos(x)  ")
print("-" * 35)

for i in range(10):
    print(f"{x[i]:<10.5f} | {sin_x[i]:<10.5f} | {cos_x[i]:<10.5f}")
```
